## Part 1: Variables and Primitives - The Building Blocks

Now we're getting to the fun part! Let's learn how computers store and use information.

### What are Variables?

Think of **variables** as labeled boxes where you can store information. Just like you might label a box "School Supplies" or "Gaming Stuff", you give variables names to remember what's inside them.

#### The Rules for Naming Variables:
* **Case sensitive** - `age` and `Age` are different variables!
* **Must start with a letter or underscore** - not a number (so `1st_place` won't work, but `first_place` will!)
* **Can only contain letters, numbers, and underscores** (A-Z, a-z, 0-9, and `_`)
* **Multi-word names should use underscores** - like `student_name` or `favorite_color`
* **Don't use [Python keywords](https://docs.python.org/3/reference/lexical_analysis.html#keywords)** - words like `print`, `for`, `if` are reserved!
* **Declaring the same variable again will overwrite it** - the old value disappears!

#### How to Create a Variable:
* Use `=` to assign a value to a variable
* Example: `age = 15` will store the number `15` in a variable called `age`

## Part 1.5: Type Casting - Transforming Data!

**Great work so far!** Now let's learn something super useful: **type casting**.

### What is Type Casting?

Type casting is like being a data transformer! Sometimes Python gives you data in one form (like a string), but you need it in another form (like a number). Type casting lets you convert between different data types.

#### Why Do We Need This?

Imagine you ask someone their age, and they say "16". To Python, that's just text (a string)! If you want to do math with it (like calculating birth year), you need to convert it to a number first.

#### The Main Casting Functions:

* `int()` - Converts to an integer (whole number)
* `float()` - Converts to a floating-point number (decimal)
* `str()` - Converts to a string (text)
* `bool()` - Converts to a boolean (True/False)

Let's see them in action!

````python
# Converting strings to numbers
age_text = "16"
print("This is a string:", age_text)
print("Type:", type(age_text))

age_number = int(age_text)  # Convert string to integer
print("\nNow it's a number:", age_number)
print("Type:", type(age_number))

# Now we can do math with it!
birth_year = 2025 - age_number
print("You were born around:", birth_year)

````

````python
# Converting numbers to strings
score = 95
lives = 3

# Try to combine them - this will cause an error! (Try uncommenting the line below)
# message = "You scored " + score + " points with " + lives + " lives left"

# The right way - convert numbers to strings first:
message = "You scored " + str(score) + " points with " + str(lives) + " lives left"
print(message)

````

````python
# Converting between int and float
whole_number = 42
decimal_number = 3.14159

# int to float
print("Integer to float:", float(whole_number))  # Output: 42.0

# float to int (WARNING: This removes the decimal part!)
print("Float to integer:", int(decimal_number))  # Output: 3

# Be careful - you lose the decimal part!
price = 19.99
print("Price as integer:", int(price))  # Output: 19 (loses the .99!)

````

#### The `input()` Problem

Here's something **really important**: When you use `input()` to get data from a user, Python **always** treats it as a string - even if they type a number!

This is one of the most common beginner mistakes. Let's see why type casting is crucial:

````python
# THE WRONG WAY (don't do this!)
# Uncomment these lines to see the error:
# num1 = input("Enter first number: ")
# num2 = input("Enter second number: ")
# result = num1 + num2  # This will concatenate strings, not add numbers!
# print("Result:", result)  # If you enter 5 and 3, you get "53" instead of 8!

print("If you entered 5 and 3, you'd get '53' instead of 8!")

````

````python

# THE RIGHT WAY (do this instead!)
num1 = int(input("Enter first number: "))  # Convert to int immediately!
num2 = int(input("Enter second number: "))
result = num1 + num2  # Now this actually adds the numbers
print("Result:", result)  # 5 + 3 = 8

````

## Part 2.5: Strings - Playing with Operators

Strings are just another data type - and just like numbers, they have their own operators! Some operators you already know behave a little differently when you use them on text instead of numbers.

### String Operators

* `+` does **concatenation** (joining strings together)
* `*` does **repetition** (repeating a string multiple times)
* `==` and `!=` check if two strings are **equal** or **not equal** (remember, this is case sensitive!)

Note: strings do **not** support `-`, `/`, `//`, `%`, or `**` - those will throw an error!

````python
# Concatenation with +
greeting = "Hello" + ", " + "Coding Cafe" + "!"
print(greeting)

# Repetition with *
border = "-" * 20
print(border)
````

````python
# Comparison with == and !=
print("apple" == "apple")   # True
print("apple" == "Apple")   # False - case sensitive!
print("apple" != "banana")  # True
````
